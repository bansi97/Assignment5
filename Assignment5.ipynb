```python
#Q1. Write a Python function count_vowels that accepts a string and returns the number of vowels a, e, i, o, u, A, E, I, O, U in that string.
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for letter in text:
        if letter in vowels:
            count += 1

    return count


result = count_vowels("Welcome")
print(result, "vowels")
```

```python
#Q3. Write a Python function display_prices that takes a product name and an arbitrary number of keyword arguments representing product prices.
#The function should display each product and its price.
def display_prices(product, **prices):
    print("Product:", product)

    for store, price in prices.items():
        print(f"{store}: ${price}")


display_prices(
    "Laptop",
    Amazon=800,
    eBay=750,
    Walmart=780
)
```

```python
#Q6.Write a Python function calculate_rectangle that accepts the length and width of a rectangle as arguments. Inside this function, create two nested functions:
# calculate_area: This function should calculate the area of the rectangle.
#calculate_perimeter: This function should calculate the perimeter of the rectangle.
#The main function should return both the area and the perimeter of the rectangle.
def calculate_rectangle(length, width):

    def calculate_area():
        return length * width

    def calculate_perimeter():
        return 2 * (length + width)

    area = calculate_area()
    perimeter = calculate_perimeter()

    return area, perimeter


area, perimeter = calculate_rectangle(5, 3)

print("Area:", area)
print("Perimeter:", perimeter)

```

```python
#Q8. Write a nested function to calculate the total hotel bill based on the number of nights.
#A hotel charges Rs. 5,000 per night. The customer must also pay 10% tax on the room cost

def calculate_hotel_bill(nights):

    def calculate_room_cost():
        return nights * 5000

    def calculate_tax():
        return calculate_room_cost() * 0.10

    room_cost = calculate_room_cost()
    tax = calculate_tax()
    total_bill = room_cost + tax

    return room_cost, tax, total_bill


room_cost, tax, total_bill = calculate_hotel_bill(3)

print("Room Cost:", room_cost)
print("Tax:", tax)
print("Total Bill:", total_bill)
```

```python
#Q9.User Login Validation
#A website checks a user's username and password before allowing access. Write a nested function to validate the login details
def login(username, password):

    def validate_username():
        return username == "admin"

    def validate_password():
        return password == "python123"

    if validate_username() and validate_password():
        return "Login Successful! Welcome."
    else:
        return "Invalid username or password. Try again."


# Calling the main function
result = login("admin", "python123")

print(result)
```

```python
#Q10.Write a Python function shopping_bill(price, quantity, discount=0) that uses nested functions:

# An inner function calculate_total() → multiplies price * quantity.
# An inner function apply_discount(total) → subtracts discount from total (but never below 0).

def shopping_bill(price, quantity, discount=0):

    def calculate_total():
        return price * quantity

    def apply_discount(total):
        return max(0, total - discount)

    total = calculate_total()
    final_bill = apply_discount(total)

    return final_bill


# Without discount
print("Final Bill =", shopping_bill(200, 3), "PKR")

# With discount
print("Final Bill =", shopping_bill(200, 3, 100), "PKR")
```
